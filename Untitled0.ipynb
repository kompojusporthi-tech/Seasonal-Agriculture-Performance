```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('seasonal_agriculture_performance_dataset.csv')

# Show basic overview
print("Dataset Shape (Rows, Columns):", df.shape)
print("\nFirst 5 rows of data:")
df.head()
```

```python
# Check for missing values and summary statistics
print("Missing values in each column:")
print(df.isnull().sum())

print("\nDataset Summary Statistics:")
df.describe()
```

```python
# 1. Calculate Average Yield and Profit by Season
seasonal_summary = df.groupby('Season')[['Yield_Tonnes_Ha', 'Profit_INR', 'Production_Tonnes']].mean().reset_index()
print("Seasonal Averages:")
print(seasonal_summary)

# 2. Generate Seasonal Comparison Graphs
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Yield by Season
sns.barplot(data=df, x='Season', y='Yield_Tonnes_Ha', ax=axes[0], palette='Greens')
axes[0].set_title('Average Crop Yield (Tonnes/Ha) by Season')
axes[0].set_ylabel('Yield (Tonnes/Ha)')

# Plot 2: Profit by Season
sns.barplot(data=df, x='Season', y='Profit_INR', ax=axes[1], palette='Blues')
axes[1].set_title('Average Profit (INR) by Season')
axes[1].set_ylabel('Profit (INR)')

plt.tight_layout()
plt.show()
```
